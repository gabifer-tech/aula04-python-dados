### Tutorial: Estrutura `match-case` no Python

A partir do Python 3.10, foi introduzida a estrutura `match-case`, que funciona de forma semelhante ao `switch-case` de outras linguagens, mas com recursos muito mais poderosos chamados de *Pattern Matching* (Correspondência de Padrões).

#### Sintaxe Básica

```python
match variavel:
    case valor_1:
        # Código executado se variavel == valor_1
    case valor_2:
        # Código executado se variavel == valor_2
    case _:
        # O caractere '_' serve como padrão (default), capturando qualquer outro valor
```

#### Exemplo Super Simples: Cor Favorita

Aqui está um exemplo simples de correspondência de strings para você entender como a escolha direta é feita:

````python
# Um exemplo muito simples com cores
cor = "azul"

match cor:
    case "vermelho":
        print("Você escolheu vermelho, a cor da paixão!")
    case "azul":
        print("Você escolheu azul, a cor do céu!")
    case "verde":
        print("Você escolheu verde, a cor da natureza!")
    case _:
        print("Cor desconhecida!")
````

````python
# Exemplo 1: Correspondência Simples (Valores Exatos)
def verificar_status(codigo_status):
    match codigo_status:
        case 200:
            return "OK"
        case 404:
            return "Não Encontrado"
        case 500:
            return "Erro Interno do Servidor"
        case _:
            return "Status Desconhecido"

# Testando o exemplo
print(verificar_status(200))
print(verificar_status(404))
print(verificar_status(999))
````

#### Casos Múltiplos e Condicionais

Podemos agrupar múltiplos padrões em um único `case` utilizando o operador `|` (OR), e também adicionar condições extras chamadas de *guards* usando a palavra-chave `if`.

````python
# Exemplo 2: Casos Múltiplos e Guards (Condições adicionais)
def classificar_dia(dia, fim_de_semana_proximo=False):
    match dia:
        case "Sábado" | "Domingo":
            return "Fim de semana! Aproveite!"
        case "Sexta" if fim_de_semana_proximo:
            return "Sextou! O fim de semana já começou."
        case "Segunda" | "Terça" | "Quarta" | "Quinta" | "Sexta":
            return "Dia útil de trabalho/estudo."
        case _:
            return "Dia inválido!"

print(classificar_dia("Sábado"))
print(classificar_dia("Sexta", fim_de_semana_proximo=True))
print(classificar_dia("Quarta"))
````

````python
# Exemplo comparativo: Resolvendo o mesmo problema com as duas abordagens

animal = "gato"

# 1) Abordagem clássica com IF, ELIF e ELSE
print("--- Usando IF-ELIF-ELSE ---")
if animal == "cachorro" or animal == "gato":
    print("É um animal de estimação doméstico!")
elif animal == "leão" or animal == "tigre":
    print("É um felino selvagem!")
else:
    print("Outro tipo de animal.")

# 2) Mesma lógica usando MATCH-CASE
print("\n--- Usando MATCH-CASE ---")
match animal:
    case "cachorro" | "gato":
        print("É um animal de estimação doméstico!")
    case "leão" | "tigre":
        print("É um felino selvagem!")
    case _:
        print("Outro tipo de animal.")
````
